```python
import requests
from bs4 import BeautifulSoup
```

```python
# sent request to webpagedata_anaylst_skills
url="https://m.timesjobs.com/mobile/jobs-search-result.html?txtKeywords=Data+Analysis%2C&cboWorkExp1=-1&txtLocation="

response=requests.get(url)
print(response)
```

```python
soup=BeautifulSoup(response.text,'html.parser')
```

```python
titles=soup.find_all('h2')
for title in titles:
    print(title.text)
```

```python
jobs_name=soup.find_all('h3')
for y in jobs_name:
     print(y.text)
```

```python
companies=soup.find_all('span',class_='srp-comp-name')
for company in companies:
    print(company.text)
```

```python
data_anaylst_skills=soup.find_all('div',class_="srp-keyskills")
for skills in data_anaylst_skills:
    print(skills.text)
```

```python
jobs_location=soup.find_all("div",class_="srp-loc")
for loc in jobs_location:
    print(loc.text)
```

```python
from os import write
import csv

header=["company","jobs","location","skills"]

# Ensure all lists have the same length by taking the minimum length
min_len = min(len(companies), len(jobs_name), len(jobs_location), len(data_anaylst_skills))

# Extract text from the BeautifulSoup objects and create a list of lists
data = []
for i in range(min_len):
    data.append([
        companies[i].text.strip() if i < len(companies) else "",
        jobs_name[i].text.strip() if i < len(jobs_name) else "",
        jobs_location[i].text.strip() if i < len(jobs_location) else "",
        data_anaylst_skills[i].text.strip() if i < len(data_anaylst_skills) else ""
    ])

with open("world_jobs.csv","w",encoding="utf-8", newline='') as f:
    writer=csv.writer(f)
    writer.writerow(header)
    writer.writerows(data)

print("CSV file 'world_jobs.csv' created successfully.")
```
